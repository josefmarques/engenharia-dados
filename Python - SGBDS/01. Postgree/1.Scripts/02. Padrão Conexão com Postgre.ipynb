```python
import sys
print(sys.executable)

import psycopg2
print(psycopg2.__version__)

dbname   = 'postgres'
user     = 'zemarques'
password = 'mrq831028'
host     = 'localhost'
port     = '5432' 

try:
    conn = psycopg2.connect(dbname=dbname,
                            user=user,
                            password=password,
                            host=host,
                            port=port)
    print("✅ Conexão estabelecida com sucesso!")

    cur = conn.cursor()

    # Consulta de teste
    cur.execute("SELECT version();")
    versao = cur.fetchone()
    print("Versão do PostgreSQL:", versao[0])

    conn.commit()
    conn.close()
    print("✅ Conexão fechada corretamente.")

except Exception as e:
    print("❌ Erro:", e)
```
